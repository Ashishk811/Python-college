 # Experiment 7(A): Making Http Requests (Client-side) 

```python
import requests
```

### GET Requests

```python
res = requests.get("https://google.com/")
print(res.text[:700])
print("-"*50)
print(res.status_code)
```

### POST Requests

```python
payload = {'username':'user123','password':'password123'}
res = requests.post("https://google.com/",json=payload)
print(res.json())
```

```python
URL='https://www.google.com/'
res = requests.get(URL)
print(f"Status code: {res.status_code}")

print("Request was successful") if res.ok else print("Request Failed")

print(f"Response Text: {res.text[:100]}--->>>") # print first 100 characters

print(f"Response Headers: {res.headers}")
```
