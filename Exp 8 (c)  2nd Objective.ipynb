# Experiment 8 (a): Implementation UDP in Python

```python
import socket
```

```python
TARGET_HOST ="www.google.com"
TARGET_PORT = 12345  
```

```python
try:
    udp_socket = socket.socket(socket.AF_INET, socket.SOCK_DGRAM)
    print("UDP socket created successfully.")
except socket.error as e:
    print(f"Error creating socket: {e}")
    exit()


message = b"Hello from Python UDP client!" # b prefix indicates bytes

try:
    # Resolve the IP address of the target host
    target_ip = socket.gethostbyname(TARGET_HOST)
    print(f"Resolved IP address for {TARGET_HOST}: {target_ip}")

    # Send the datagram packet
    udp_socket.sendto(message, (target_ip, TARGET_PORT))
    print(f"Datagram packet sent to {TARGET_HOST}:{TARGET_PORT}")

except socket.gaierror as e:
    print(f"Error resolving host {TARGET_HOST}: {e}")
except socket.error as e:
    print(f"Error sending data: {e}")

finally:
    udp_socket.close()
    print("UDP socket closed.")
```
