# Experiment 8 (a): Implementation UDP in Python

### Sending UDP Packets 

```python
import socket
```

```python
UDP_IP= "127.0.0.1"
UDP_PORT= 5001
```

```python
# Create a UDP socket
sock= socket.socket(socket.AF_INET, # Internet socket IPv4
                    socket.SOCK_DGRAM) # UDP protocol

message= "Hello, UDP!"

try:
    sock.sendto(message.encode(),(UDP_IP,UDP_PORT))
    print(f"Sent UDP packet to {UDP_IP}: {UDP_PORT}: {message}")
    
finally:
    sock.close()
```
